```python
from matplotlib import pyplot as plt
from matplotlib.lines import lineStyles

fig =plt.figure(figsize=(15,8),dpi=80)#dpi是每英寸上像素点的个数
x = range(2,26,2)# start,end+1,step(不能为小数)
y = [13,23,43,32,34,32,43,24,13,43,35,36]
plt.xticks(range(2,26,2))
plt.yticks(range(min(y),max(y)+1,2))#记得加一
plt.plot(x,y) #draw
plt.savefig("./picture.svg")#保存在本文件的地址下,可以保存为svg的矢量图形式，放大无锯齿
plt.show() #show
```

```python
from matplotlib import pyplot as plt
import matplotlib
import random

plt.rcParams['font.sans-serif'] = ['Microsoft YaHei']
plt.rcParams['axes.unicode_minus'] = False     #


fig =plt.figure(figsize=(33,8),dpi=80)#dpi是每英寸上像素点的个数
x = list(range(0,120,1))#不是在外面打【】，如果不切片就不用list
y_1 = [random.randint(20,25) for i in range(120)]#for前面没有，;外面有一层【】
y_2 = [random.randint(20,25) for i in range(120)]#for前面没有，;外面有一层【】
#若想刻度有小数可以用下列代码
# x_label = [i/2 for i in range(4,49)]
#若想刻度有字符串
x_labels = ["10点{}分".format(i) for i in range(60)]
x_labels += ["11点{}分".format(i-60) for i in range(60,120)]#用+=。因为= 是重新绑定，+= 是原地追加（对列表来说等价于 x_labels.extend([...])）
plt.xticks(x[::3],x_labels[::3],rotation = 45)
#这一步把数字和字符串一一对应
#rotation是刻度旋转多少度的意思
plt.yticks(range(min(y),max(y)+1,2))#记得加一

plt.plot(x,y_1,label = "1",color = "orange",linestyle=':',linewidth = 0.5)#冒号要用英文的！！！！s不能大写,linewidth别写错
plt.plot(x,y_2,label = "2",color = "red",linestyle = "-.")#画两个


plt.xlabel("时间")
plt.ylabel("温度")
plt.title("温度-时间")
plt.grid(alpha = 0.4)#按刻度画网格图,alphy是透明度
plt.legend(loc = 0)#加图例,loc是图例所在位置（详情见下）

plt.savefig("./picture2.svg")
plt.show()
```

# matplotlib 四种基础线型

| 符号写法 | 全名写法 | 效果   | 说明                       |
| -------- | -------- | ------ | -------------------------- |
| `'-'`    | `'solid'`  | 实线   | 默认样式，不写就是它       |
| `'--'`   | `'dashed'` | 虚线   | 长横线 + 等宽空隙          |
| `'-.'`   | `'dashdot'`| 点划线 | 长横线 + 点，交替出现      |
| `':'`    | `'dotted'` | 点线   | 密集小圆点，视觉最轻       |

# plt.legend(loc=...) 图例位置速查

数字和字符串等价：loc=0 == loc='best'。不写 loc 默认就是 0（best）。

| 数字 | 字符串           | 位置       |
| ---- | ---------------- | ---------- |
| 0    | `'best'`         | 自动挑最不挡曲线的位置（默认） |
| 1    | `'upper right'`  | 右上       |
| 2    | `'upper left'`   | 左上       |
| 3    | `'lower left'`   | 左下       |
| 4    | `'lower right'`  | 右下       |
| 5    | `'right'`        | 右侧居中   |
| 6    | `'center left'`  | 左侧居中   |
| 7    | `'center right'` | 右侧居中   |
| 8    | `'lower center'` | 底部居中   |
| 9    | `'upper center'` | 顶部居中   |
| 10   | `'center'`       | 正中央     |

记忆规律：位置是按数字顺时针排的——1 右上起，沿着边框转一圈（2 左上 → 3 左下 → 4 右下），5~9 是四边和侧边的中点，10 是正中心，0 是让 matplotlib 自己挑。
